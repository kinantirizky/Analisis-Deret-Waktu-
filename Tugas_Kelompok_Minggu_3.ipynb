```python
plt.figure()
plt.plot(ma2)
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.show()
```

Berikut merupakan scatter plot antara nilai pengamatan saat ini dengan nilai pengamatan t-1 (lag 1), t-2, (lag 2), dan t-3 (lag 3).

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma2), lag=1, c='k')
plt.xlabel(r'$Y_{t-1}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

<h2> AR (2) </h2>

Proses AR (2) memiliki persamaan:

<center> $Y_t=\phi_1 Y_{t-1}+\phi_2 Y_{t-2}+e_t$ </center>

**Syarat kestasioneran untuk deret AR (2) adalah jika dan hanya jika tiga kondisi berikut terpenuhi, yaitu:**

**<center> $\phi_1+\phi_2<1$  ,   $\phi_2-\phi_1<1$ ,   $|\phi_2|<1$ </center>**

Kita simulasikan deret AR (2) dengan $\phi_1=1.5$ dan $\phi_2=-0.05$

```python
np.random.seed(191)
phi1_ar2, phi2_ar2 = 1.5, -0.05

ar_params = np.array([1, -phi1_ar2, -phi2_ar2]) #nilai parameter theta dalam sintaks dituliskan sebagai negatif dari nilai parameter asli
ma_params = np.array([1])

ar_object2 = ArmaProcess(ar_params, ma_params)
ar2_s = ar_object2.generate_sample(nsample=100)

plt.figure(figsize=(10,5))
plt.plot(ar2_s,'-o')
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.title(f'Simulasi Data AR(2)')
plt.show()
```

```python
plot_acf(np.array(ar2_s), zero=False)
plt.show()
```

```python
plot_pacf(np.array(ar2_s), zero=False)
plt.show()
```
